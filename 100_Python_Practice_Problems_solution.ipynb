### 15. Overlapping rectangles.
**Example Input:** `L1=(0,10), R1=(10,0), L2=(5,5), R2=(15,-5)`  
**Example Output:** `True (Overlap)`

```
# Write your code here
# two rectangles do not overlap if either of these conditions is true:
# One rectangle is completely to the left of the other:
# R1.x < L2.x or R2.x < L1.x
# One rectangle is completely above the other:
# L1.y < R2.y or L2.y < R1.y

L1, R1, L2, R2 = (0,10), (10,0), (5,5), (15,-5)

if (R1[0] < L2[0] or R2[0] < L1[0])  or  (L1[1] < R2[1] or L2[1] < R1[1]):
  print("they do no overlap")
else:
  print("they are overlapping")
```

### 33. User will provide 2 numbers you have to find the LCM of those 2 numbers
**Example Input:** `4, 5`  
**Example Output:** `20`

```
# Write your code here
from functools import reduce

numbers = [4, 5]
result = []

factor = 2

while True:

    # stop when all numbers become 1
    if all(num == 1 for num in numbers):
        break

    divided = False

    for index in range(len(numbers)):

        if numbers[index] % factor == 0:
            numbers[index] = numbers[index] // factor
            divided = True

    if divided:
        result.append(factor)
    else:
        factor += 1

print("Factors:", result)

lcm = reduce(lambda x, y: x * y, result)

print("LCM:", lcm)


```

Compound Interest:

**Example Output:** `2`

```
# Write your code here
def find_frequency_of(char, sentence):
  a=dict()
  for i in sentence:
    if i not in a:
      a[i]=1
    else:
      a[i]+=1

  if char in a.keys():
    return a[char]

sentence='hello how are you'
char='h'
print(find_frequency_of(char,sentence))

```

```
# Write your code here
def union(l1,l2):
  result=[]
  for i in l1:
    result.append(i)

  for j in l2:
    if j not in l1:
      result.append(j)
  return result

def intersection(l1,l2):
  result=[]
  for i in l1:
    if i in l2:
      result.append(i)
  return result

l1=[1,2,3,7]
l2=[2,3,4,6]
print(union(l1,l2))
print(intersection(l1,l2))
```

### 77. Sort a given unsorted list without using any built in function.
**Example Input:** `[3, 1, 2]`  
**Example Output:** `[1, 2, 3]`

```
# Write your code here
# i am using insertion sort algorithm there are a lot of algorithms i find it easy to remember this that's why i am using this you can try merge sort for better speed
def insertion_sort(L):
  n = len(L)
  if n<1:
    return L

  for i in range(n):
    j=i
    while ( j>0 and L[j] < L[j-1] ):
      (L[j],L[j-1]) = (L[j-1],L[j])
      j=j-1
  return L

l1=[3,2,5,1,8]
print(insertion_sort(l1))
```

### 88. Write a program to flatten a nested dictionary.
**Example Input:** `{'a': {'b': 1}}`  
**Example Output:** `{'a_b': 1}`

```
# Write your code here
def flatten_dictionary(dictionary, parent_key='', separator='_'):
    items = []
    for key, value in dictionary.items():
        new_key = parent_key + separator + key if parent_key else key
        if isinstance(value, dict):
            items.extend(flatten_dictionary(value, new_key, separator=separator).items())
        else:
            items.append((new_key, value))
    return dict(items)

nested_dict = {'a': {'b': 1}}
flattened_dict = flatten_dictionary(nested_dict)
print(flattened_dict)
```

### 98. Write a program to find the power of a number using recursion.
**Example Input:** `base=2, exp=3`  
**Example Output:** `8`

```
# Write your code here
def power(base, exp):
  if exp == 0:
    return 1
  elif exp > 0:
    return base * power(base, exp - 1)
  else:
    return 1 / (base * power(base, -exp - 1))

base = 2
exp = 3
print(power(base, exp))
```

### 100. Write a program to find the longest consecutive sequence in an unsorted array.
**Example Input:** `[100, 4, 200, 1, 3, 2]`  
**Example Output:** `4 (Sequence: [1, 2, 3, 4])`

```
# Write your code here
def longest_consecutive_sequence(nums):
    if not nums:
        return 0

    num_set = set(nums)
    max_len = 0

    for num in num_set:
        if (num - 1) not in num_set: # Check if it's the start of a sequence
            current_num = num
            current_streak = 1

            while (current_num + 1) in num_set:
                current_num += 1
                current_streak += 1

            max_len = __builtins__.max(max_len, current_streak) # Explicitly call built-in max

    return max_len

nums = [100, 4, 200, 1, 3, 2]
print(longest_consecutive_sequence(nums))
```
